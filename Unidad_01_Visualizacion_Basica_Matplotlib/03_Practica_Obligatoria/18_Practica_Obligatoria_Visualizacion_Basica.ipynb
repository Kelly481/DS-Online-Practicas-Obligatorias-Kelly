#### #1.2  

Recrea la siguiente figura, para ello construye en apartados separados cada gráfico y luego haz un apartado en el que se junten los cuatro.  

Las figuras son:
1. Las ciudades más pobladas, con los numeros de las poblaciones (tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento `ha` a "center"). (Colores: azul, gris, aceituna, marrón, con trasparencia)  

2. Histograma del área total ocupada por las ciudades. (50 bins, fuente = "Bradley Hand ITC", tamaño 18, color azul acero o "steelblue")
3. El scatter de área de agua de un ciudad con su poblacion.
4. El boxplot de la altura de las ciudades.

```python
# Ciudades mas pobladas

df = pd.read_csv("./data/california_cities.csv")

top = df.nlargest(4, "population_total")

colores = ["blue", "grey", "olive", "brown"]

plt.figure(figsize=(8,5))
plt.bar(top["city"], top["population_total"], color=colores, alpha=0.7)

for i, v in enumerate(top["population_total"]):
    plt.text(i, v, str(int(v)), ha="center", va="bottom")

plt.title("Ciudades más pobladas")
plt.ylabel("Población")
plt.show()

```

```python
# Histograma

plt.figure(figsize=(8,5))
plt.hist(df["area_total_sq_mi"], bins=50, color="steelblue")

plt.title("Histograma del área total", fontname="Bradley Hand ITC", fontsize=18)
plt.xlabel("Área total (sq mi)")
plt.ylabel("Frecuencia")
plt.show()
```

```python
# scatter de área de agua de un ciudad con su poblacion.

plt.figure(figsize=(8,5))
plt.scatter(df["area_water_sq_mi"], df["population_total"], alpha=0.7)

plt.title("Área de agua vs población")
plt.xlabel("Área de agua (sq mi)")
plt.ylabel("Población")
plt.show()
```

```python
# Boxplot de la elevación

plt.figure(figsize=(6,5))
plt.boxplot([df["elevation_m"].dropna()])  # pasar como lista y sin NaN
plt.title("Distribución de la elevación de las ciudades", fontsize=12)
plt.ylabel("Elevación (m)", fontsize=11)


```

```python
# Cerrar figuras previas
plt.close("all")

# Preparar datos

top = df.nlargest(4, "population_total")
colores = ["blue", "grey", "olive", "brown"]

area = df["area_total_sq_mi"].dropna()
x = df["area_water_sq_mi"]
y = df["population_total"]
elev = df["elevation_m"].dropna()


# Figura general

fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# SUBPLOT 1 — Ciudades más pobladas

axs[0,0].bar(top["city"], top["population_total"], color=colores, alpha=0.7)
for i, v in enumerate(top["population_total"]):
    axs[0,0].text(i, v + 50, f"{v:.0f}", ha="center")
axs[0,0].set_title("Ciudades más pobladas")
axs[0,0].set_ylabel("Habitantes")

# SUBPLOT 2 — Histograma

axs[0,1].hist(area, bins=50, color="steelblue")
axs[0,1].set_title("Distribución del área total ocupada")
axs[0,1].set_xlabel("Área total (sq mi)")
axs[0,1].set_ylabel("Frecuencia")

# SUBPLOT 3 — Scatter

axs[1,0].scatter(x, y, alpha=0.6)
axs[1,0].set_title("Área de agua vs Población")
axs[1,0].set_xlabel("Área de agua (sq mi)")
axs[1,0].set_ylabel("Población total")

# SUBPLOT 4 — Boxplot

axs[1,1].boxplot([elev])
axs[1,1].set_title("Distribución de la elevación")
axs[1,1].set_ylabel("Elevación (m)")

plt.tight_layout()


```
